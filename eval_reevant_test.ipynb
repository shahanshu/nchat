```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create the new dataset
data = {
    'Test': ['test1', 'test2', 'test3', 'test4', 'test5', 'test6', 'test7', 'test8', 'test9', 'test10'],
    'Ans Relevance': [89, 91, 90, 84, 83, 87, 88, 85, 92, 86],
    'Ret Relevance': [40, 49, 55, 59, 49, 45, 52, 47, 58, 50],
    'Groundedness': [64, 63, 57, 64, 64, 68, 66, 62, 69, 65],
    'Correctness': [91, 92, 86, 91, 86, 92, 90, 89, 93, 88]
}

df = pd.DataFrame(data)

# 2. Reshape the data for Seaborn (melt from wide to long format)
df_melted = df.melt(id_vars='Test', var_name='Metric', value_name='Score (%)')

# 3. Set up the Seaborn visualization style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# 4. Create the line plot with markers (symbols)
ax = sns.lineplot(
    data=df_melted, 
    x='Test', 
    y='Score (%)', 
    hue='Metric',       # Colors the lines by metric
    style='Metric',     # Gives each metric a unique symbol
    markers=True,       # Turns on the symbols
    dashes=False,       # Keeps lines solid
    markersize=10,      # Makes the symbols larger
    linewidth=2
)

# 5. Customize the chart labels, title, and limits
plt.title('Evaluation Metrics For Irrelevant Queries/web search', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Test Case', fontsize=12, fontweight='bold')
plt.ylabel('Score (%)', fontsize=12, fontweight='bold')

# Adjusted Y-axis to accommodate the lower Retrieval and Groundedness scores
plt.ylim(35, 100) 

# 6. Format the Legend (placed outside the plot)
plt.legend(
    title='Evaluation Metrics', 
    title_fontsize='11', 
    bbox_to_anchor=(1.02, 1), 
    loc='upper left', 
    borderaxespad=0.
)

# Show the plot
plt.tight_layout()
plt.savefig("irelevant_search-results.png", dpi=300, bbox_inches='tight')
plt.show()
```
