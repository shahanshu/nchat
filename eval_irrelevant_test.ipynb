# --- Cell 1 ---
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create the dataset
data = {
    'Test': ['test1', 'test2', 'test3', 'test4', 'test5', 'test6', 'test7', 'test8', 'test9', 'test10'],
    'Ans Relevance': [98, 87, 98, 91, 91, 91, 86, 86, 95, 98],
    'Ret Relevance': [90, 89, 94, 80, 92, 87, 94, 85, 100, 82],
    'Groundedness': [98, 93, 90, 90, 86, 90, 98, 87, 100, 95],
    'Correctness': [93, 98, 92, 96, 92, 95, 96, 98, 96, 93]
}

df = pd.DataFrame(data)

# 2. Reshape the data for Seaborn (melt from wide to long format)
# This makes it easy to plot multiple metrics with a legend and symbols
df_melted = df.melt(id_vars='Test', var_name='Metric', value_name='Score (%)')

# 3. Set up the Seaborn visualization style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6)) # Make the chart wide enough to fit all tests easily

# 4. Create the line plot with markers (symbols)
ax = sns.lineplot(
    data=df_melted, 
    x='Test', 
    y='Score (%)', 
    hue='Metric',       # Colors the lines by metric
    style='Metric',     # Gives each metric a unique symbol (marker)
    markers=True,       # Turns on the symbols
    dashes=False,       # Keeps lines solid rather than dashed
    markersize=10,      # Makes the symbols larger and easier to read
    linewidth=2
)

# 5. Customize the chart labels, title, and limits
plt.title('Evaluation Metrics Performance Across Tests', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Test Case', fontsize=12, fontweight='bold')
plt.ylabel('Score (%)', fontsize=12, fontweight='bold')
plt.ylim(75, 105) # Adjusting Y-axis to focus on the 80-100% range where your data lives

# 6. Format the Legend
# Place legend outside the plot so it doesn't cover any data lines or symbols
plt.legend(
    title='Evaluation Metrics', 
    title_fontsize='11', 
    bbox_to_anchor=(1.02, 1), 
    loc='upper left', 
    borderaxespad=0.
)

# Show the plot
plt.tight_layout()
plt.savefig("irelevant_search-results.png", dpi=300, bbox_inches='tight')
plt.show()
